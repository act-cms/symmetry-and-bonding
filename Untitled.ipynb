```python
import numpy as np

water = np.array([
  ['atom', 'x',  'y',  'z'],
  ['O',     0,    0,     0],
  ['H',     0,    1,     1],
  ['H',     0,   -1,    -1],
  #['D',     1,    1,     1],
])

rot_mat = np.array([
    [1,  0,  0],
    [0,  0,  1],
    [0, -1,  0],
])

water_xyz = np.array([x[1:] for x in water[1:]]).astype(int)
print(water)
print(water_xyz)
print(rot_mat)

water_xyz_rot = np.matmul(water_xyz, rot_mat)
print(water_xyz_rot)
```
