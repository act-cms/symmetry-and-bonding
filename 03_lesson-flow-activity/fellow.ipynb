# ACT-CMS Lesson Flow
---


The purpose of this hands-on working session is for ACT-CMS Fellows to develop a familiarity with the technical aspects of developing their proposed lesson(s), including JupyterLab, ChemCompute, and the Lesson Flow version control framework.

## Session Information

### Session Learning Outcomes

At the end of this session, ACT-CMS Fellows will be able to...

1. Use GitHub to create a new lesson repository within the ACT-CMS organization from the `act-cms/lesson-cookiecutter` template repository (referred to below as `act-cms/lesson`);
2. Use `git`/GitHub Desktop to make a local clone of their lesson repository (referred to below as `local/lesson`);
3. Use `git`/GitHub Desktop to navigate the branches of `local/lesson`;
4. Use JupyterLab to customize the activity template by developing part of a Python basics lesson for their target course;
5. Use `git`/GitHub Desktop to diff, stage, and commit local changes to the activity template to the branch `local/lesson:contrib`;
6. Use `git`/GitHub Desktop to push their local commit(s) from `local/lesson:contrib` to `act-cms/lesson:contrib`

### Cyberinfrastructure Prerequisites

Before beginning this session, Fellows are expected to...

1. Complete the setup of their local machine according to the steps outlined
in [Chapter 2: Setup for ACT-CMS Fellows](02_fellow-setup.md)
2. Be familiar with basic Python programming and syntax, at the level of the
[Introduction to Programming for Molecular
Scientists](https://act-cms.molssi.org/portal/lessons/foundational-intro-python/)
lesson on the ACT-CMS Portal

There are no content prerequisites for this session.

## I. Session Setup


Before launching this notebook, the following steps will have already been completed by each Fellow:

1. Create a lesson repository on GitHub (`act-cms/lesson`)
2. Clone `act-cms/lesson` to your personal computer (`local/lesson`)
3. Tour branches on both `act-cms/lesson` and `local/lesson` via GH and `git`/GHD; land on `local/lesson:contrib` branch
4. Python installation & environment setup
    * Install Python from Anaconda Distribution
    * Build `chemcomp2.5` environment from `environment.yml` file
    * Activate `chemcomp2.5`
5. Launch JupyterLab from `local/lesson:contrib`
6. Open this file, `fellow.ipynb`

## II. Activity Planning & Instructions

### II.A. Overview


Now that you have set up your local environment and are ready for development, you will work collaboratively in small mentor/mentee groups to write a _short_ activity introducing a _single topic in basic Python syntax_ (e.g., loops, list comprehensions, writing functions, etc.) to the students in the target course from your proposal. This way, mentee/mentor relationships will be reinforced while everyone gains experience with using JupyterLab for lesson development as part of the Lesson Flow.
